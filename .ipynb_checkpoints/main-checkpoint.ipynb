```python
from data_preprocessing import reduce_mem_usage,expand_datetime_features,process_outliers_and_issues
from feature_engineering import count_coding,cut_group,cross_cat_num,cross_qua_cat_num,MeanTargetEncoder
from car_model import AdaptiveDropout,NN_model,Metric
from car_model import scheduler1
#from predict import predict
```

### 数据处理

```python
## 读取数据
train = reduce_mem_usage(pd.read_csv('used_car_train_20200313.csv', sep=' '))
test = reduce_mem_usage(pd.read_csv('used_car_testB_20200421.csv', sep=' '))
# 使用示例
train = process_outliers_and_issues(train)
test = process_outliers_and_issues(test)
```

```python
#异常值处理
data = pd.concat([train,test])
data = data.fillna(data.mode().iloc[0,:])
#concat_data['power'][concat_data['power']<1] = 1
```

### 特征工程

```python
#时间特征
date_cols = ['regDate', 'creatDate']
data = expand_datetime_features(data,date_cols)

# 匿名特征，加法乘法（避免重复）
for i, j in combinations(['v_' + str(i) for i in range(15)], 2):
    data[f'{i}+{j}'] = data[i] + data[j]  # 数量C(15,2)=105
for col1, col2 in combinations(['v_' + str(i) for i in range(15)], 2):
    data[f'{col1}*{col2}'] = data[col1] * data[col2] # 数量C(15,2)=105
for i in ['model','brand', 'bodyType', 'fuelType','gearbox', 'power', 'kilometer', 'notRepairedDamage', 'regionCode']:
    for j in ['v_' +str(i) for i in range(15)]:
        data[str(i)+'*'+str(j)] = data[i]*data[j]    


#count编码
count_list = ['model', 'brand', 'regionCode','bodyType','fuelType','name','regDate_year', 'regDate_month', 'regDate_day',
       'regDate_dayofweek' , 'creatDate_month','creatDate_day', 'creatDate_dayofweek','regDate', 'creatDate', 'kilometer']#
       
data = count_coding(data,count_list)

# 使用时间
data['used_time1'] = (pd.to_datetime(data['creatDate'], format='%Y%m%d', errors='coerce') - 
                            pd.to_datetime(data['regDate'], format='%Y%m%d', errors='coerce')).dt.days
data['used_time2'] = (pd.to_datetime(20200101) - pd.to_datetime(data['regDate'], format='%Y%m%d', errors='coerce')).dt.days                        
data['used_time3'] = (pd.to_datetime(20200101) - pd.to_datetime(data['creatDate'], format='%Y%m%d', errors='coerce') ).dt.days

#分桶操作
cut_cols = ['power']+['used_time1','used_time2','used_time3']
data = cut_group(data,cut_cols,50)
cut_cols = ['kilometer']+['used_time1','used_time2','used_time3']
data = cut_group(data,cut_cols,11)

```

```python
### 用数值特征对类别特征做统计刻画，随便挑了几个跟price相关性最高的匿名特征
cross_cat = ['model', 'brand','regDate_year']
cross_num = ['v_0','v_3', 'v_8', 'v_11', 'v_12','power','kilometer']
data = cross_cat_num(data,cross_num,cross_cat)#一阶交叉
data = cross_qua_cat_num(data)#二阶交叉
## 选择特征列
numerical_cols = data.columns
#print(numerical_cols)

cat_fea = ['SaleID','offerType','seller']
feature_cols = [col for col in numerical_cols if col not in cat_fea]
feature_cols = [col for col in feature_cols if col not in ['price']]

## 提前特征列，标签列构造训练样本和测试样本
X_data = data.iloc[:len(train),:][feature_cols]
Y_data = train['price']
X_test  = data.iloc[len(train):,:][feature_cols]

# 目标均值编码
categorical_features = ['model', 'brand', 'name', 'regionCode'] + date_cols
mean_encoder = MeanTargetEncoder(categorical_features, target_type='regression')

# 对训练集进行编码（带交叉验证，防止泄漏）
X_data = mean_encoder.fit_transform(X_data, Y_data)
# 对测试集进行编码（用训练集得出的编码规则）
X_test = mean_encoder.transform(X_test)

X_data['price'] = train['price']
```

```python
drop_list = ['regDate', 'creatDate','brand_power_min', 'regDate_year_power_min']
x_train = X_data.drop(drop_list+['price'],axis=1)
x_test = X_test.drop(drop_list,axis=1)
x_train.shape
```

```python
drop_list = ['regDate', 'creatDate']
x_train=X_data.drop(drop_list+['price'],axis=1)
x_test = X_test.drop(drop_list,axis=1)

x_train = x_train.astype('float32')
x_test = x_test.astype('float32')
```

```python
from sklearn.preprocessing import MinMaxScaler
from sklearn.impute import SimpleImputer
#特征归一化
min_max_scaler = MinMaxScaler()
min_max_scaler.fit(pd.concat([x_train,x_test]).values)
all_data = min_max_scaler.transform(pd.concat([x_train,x_test]).values)
imputer = SimpleImputer(strategy='mean')  # or 'median', 'most_frequent'
all_data_imputed = imputer.fit_transform(all_data)

from sklearn import decomposition
pca = decomposition.PCA(n_components=40)

all_pca = pca.fit_transform(all_data_imputed)

#all_pca = pca.fit_transform(all_data)
X_pca = all_pca[:len(x_train)]
test = all_pca[len(x_train):]
y = train['price'].values
```

### 训练

```python
from sklearn.model_selection import KFold
from tqdm import tqdm
import keras 
from keras.callbacks import LearningRateScheduler
def scheduler2(epoch):

    lr = model.optimizer.learning_rate.numpy()
    if epoch % 20 == 0 and epoch:
        model.optimizer.learning_rate.assign(lr * 0.6)
        print("lr changed to {}".format(lr * 0.6))
    return model.optimizer.learning_rate.numpy()

reduce_lr = LearningRateScheduler(scheduler2)
#reduce_lr = keras.callbacks.LearningRateScheduler(get_scheduler(model))  # 传入model
n_splits = 6
kf = KFold(n_splits=n_splits, shuffle=True)

b_size = 2048
max_epochs = 160
oof_pred = np.zeros((len(X_pca), ))

sub = pd.read_csv('used_car_testB_20200421.csv',sep = ' ')[['SaleID']].copy()
sub['price'] = 0

avg_mae = 0
for fold, (trn_idx, val_idx) in enumerate(kf.split(X_pca, y)):
    print('fold:', fold)
    X_train, y_train = X_pca[trn_idx], y[trn_idx]
    X_val, y_val = X_pca[val_idx], y[val_idx]
    
    model = NN_model(X_train.shape[1])
    simple_adam = keras.optimizers.Adam(learning_rate = 0.015)
    
    model.compile(loss='mae', optimizer=simple_adam,metrics=['mae'])
    es = EarlyStopping(monitor='val_score', patience=10, verbose=2, mode='min', restore_best_weights=True,)
    es.set_model(model)
    metric = Metric(model, [es], [(X_train, y_train), (X_val, y_val)])
    model.fit(X_train, y_train, batch_size=b_size, epochs=max_epochs, 
              validation_data = [X_val, y_val],
              callbacks=[reduce_lr], shuffle=True, verbose=2)

    y_pred3 = model.predict(X_val)
    y_pred = np.zeros((len(y_pred3), ))
    sub['price'] += model.predict(test).reshape(-1,)/n_splits
    for i in range(len(y_pred3)):
        y_pred[i] = y_pred3[i]
        
    oof_pred[val_idx] = y_pred
    val_mae = mean_absolute_error(y[val_idx], y_pred)
    avg_mae += val_mae/n_splits

    print('val_mae is:{}'.format(val_mae))
    
mean_absolute_error(y, oof_pred)
```

### lightbgm

```python
from sklearn.model_selection import KFold
from sklearn.metrics import mean_absolute_error
import numpy as np
import lightgbm as lgb
from lightgbm import early_stopping, log_evaluation, reset_parameter

def dynamic_learning_rate(boosting_round):
    """每100轮学习率衰减为原来的90%"""
    base_rate = 0.04
    min_rate = 0.000001
    decay_factor = 0.95
    lr = base_rate * (decay_factor ** (boosting_round // 250))
    return max(lr, min_rate)

X = X_pca
y = np.log1p(y)
X_test = test

# KFold 验证
kf = KFold(n_splits=6, shuffle=True, random_state=42)
preds = np.zeros(len(X_test))

for fold, (train_idx, valid_idx) in enumerate(kf.split(X), 1):
    X_train, X_valid = X[train_idx], X[valid_idx]
    y_train, y_valid = y[train_idx], y[valid_idx]
    
    model = lgb.LGBMRegressor(
        n_estimators=8000, #4000,
        learning_rate=0.04,  #0.1,
        num_leaves=128,      # 显式设置叶子数量
        max_depth=-1,        # 让num_leaves控制复杂度
        #max_depth=8,
        subsample=0.8,
        colsample_bytree=0.8,
        objective='regression_l1',   # 这里指定 MAE 目标函数
        min_split_gain=0.0 , #0.05,         # 降低分裂门槛（默认是 0.0，也可以显式设置）
        #min_child_samples=10,       # 默认 20，减小可增加叶子数量
        #reg_alpha=0.0,              # L1正则，设置为0避免惩罚过重
        #reg_lambda=0.0,             # L2正则，设置为0避免惩罚过重
        random_state=42
    )
    model.fit(
        X_train, y_train,
        eval_set=[(X_valid, y_valid)],
        eval_metric='mae',  # 验证集用 MAE 评价
        callbacks=[early_stopping(50),
        log_evaluation(period=250) , # 关闭日志输出（可设为10显示间隔输出）
        reset_parameter(learning_rate=dynamic_learning_rate)  # 动态调整           
        ]
    )
    val_pred_log = model.predict(X_valid)
    val_pred = np.expm1(val_pred_log)  # 还原
    val_true = np.expm1(y_valid)
    #val_pred = model.predict(X_valid)
    #val_mae = mean_absolute_error(y_valid, val_pred)
    val_mae = mean_absolute_error(val_true, val_pred)
    print(f"Fold {fold} 验证集 MAE: {val_mae:.4f}")
    
    #preds += model.predict(X_test) / kf.n_splits
    test_pred_log = model.predict(X_test)
    preds += np.expm1(test_pred_log) / kf.n_splits
```
